```python
from pathlib import Path
import pandas as pd

from sqlalchemy import create_engine
```

```python
BASE_DIR = Path.cwd().parent

PROCESSED_DIR = (
    BASE_DIR
    / "data"
    / "processed"
)

DB_PATH = (
    BASE_DIR
    / "data"
    / "db"
    / "bluestock_mf.db"
)

print(PROCESSED_DIR)
print(DB_PATH)
```

```python
funds = pd.read_csv(
    PROCESSED_DIR /
    "01_fund_master_clean.csv"
)

nav = pd.read_csv(
    PROCESSED_DIR /
    "02_nav_history_clean.csv"
)

performance = pd.read_csv(
    PROCESSED_DIR /
    "07_scheme_performance_clean.csv"
)

transactions = pd.read_csv(
    PROCESSED_DIR /
    "08_investor_transactions_clean.csv"
)
```

```python
summary = pd.DataFrame({
    "Dataset": [
        "Fund Master",
        "NAV History",
        "Performance",
        "Transactions"
    ],
    "Rows": [
        len(funds),
        len(nav),
        len(performance),
        len(transactions)
    ],
    "Columns": [
        funds.shape[1],
        nav.shape[1],
        performance.shape[1],
        transactions.shape[1]
    ]
})

summary
```

```python
print("FUNDS")
display(funds.isnull().sum())

print("NAV")
display(nav.isnull().sum())

print("PERFORMANCE")
display(performance.isnull().sum())

print("TRANSACTIONS")
display(transactions.isnull().sum())
```

```python
print(
    "Funds Duplicates:",
    funds.duplicated().sum()
)

print(
    "NAV Duplicates:",
    nav.duplicated().sum()
)

print(
    "Performance Duplicates:",
    performance.duplicated().sum()
)

print(
    "Transactions Duplicates:",
    transactions.duplicated().sum()
)
```

```python
engine = create_engine(
    f"sqlite:///{DB_PATH}"
)

print("Connected Successfully")
```

```python
tables = pd.read_sql(
    """
    SELECT name
    FROM sqlite_master
    WHERE type='table'
    ORDER BY name
    """,
    engine
)

tables
```

```python
pd.read_sql(
    """
    SELECT COUNT(*) AS total_funds
    FROM dim_fund
    """,
    engine
)
```

```python
pd.read_sql(
    """
    SELECT COUNT(*) AS total_nav
    FROM fact_nav
    """,
    engine
)
```

```python
pd.read_sql(
    """
    SELECT COUNT(*) AS total_performance
    FROM fact_performance
    """,
    engine
)
```

```python
pd.read_sql(
    """
    SELECT COUNT(*) AS total_transactions
    FROM fact_transactions
    """,
    engine
)
```

```python
pd.read_sql(
    """
    SELECT COUNT(*) AS total_funds
    FROM dim_fund
    """,
    engine
)
```

```python
pd.read_sql(
    """
    SELECT
        category,
        COUNT(*) AS fund_count
    FROM dim_fund
    GROUP BY category
    """,
    engine
)
```

```python
pd.read_sql(
    """
    SELECT *
    FROM fact_nav
    ORDER BY nav DESC
    LIMIT 10
    """,
    engine
)
```

```python
pd.read_sql(
    """
    SELECT
        AVG(return_3yr_pct)
        AS avg_return_3yr
    FROM fact_performance
    """,
    engine
)
```

```python
pd.read_sql(
    """
    SELECT
        amfi_code,
        sharpe_ratio
    FROM fact_performance
    ORDER BY sharpe_ratio DESC
    LIMIT 10
    """,
    engine
)
```

```python
pd.read_sql(
    """
    SELECT
        SUM(amount_inr)
        AS total_transaction_amount
    FROM fact_transactions
    """,
    engine
)
```

```python
pd.read_sql(
    """
    SELECT
        state,
        COUNT(*) AS txn_count
    FROM fact_transactions
    GROUP BY state
    ORDER BY txn_count DESC
    LIMIT 10
    """,
    engine
)
```

```python
pd.read_sql(
    """
    SELECT
        amfi_code,
        return_5yr_pct
    FROM fact_performance
    ORDER BY return_5yr_pct DESC
    LIMIT 10
    """,
    engine
)
```

```python
pd.read_sql(
    """
    SELECT
        AVG(nav)
        AS avg_nav
    FROM fact_nav
    """,
    engine
)
```

```python
pd.read_sql(
    """
    SELECT
        state,
        SUM(amount_inr)
        AS total_amount
    FROM fact_transactions
    GROUP BY state
    ORDER BY total_amount DESC
    """,
    engine
)
```

```python
engine.dispose()

print("Database Connection Closed")
```
