## Notebook 01 - Data Ingestion

### Objectives

- Load all 10 raw datasets
- Understand dataset structure
- Check missing values
- Check duplicates
- Perform initial profiling

```python
from pathlib import Path
import pandas as pd
import os
```

```python
BASE_DIR = Path.cwd().parent

folder = BASE_DIR / "data" / "raw"

print(f"Reading files from: {folder}")
```

## Dataset Inventory

The project contains 10 mutual fund datasets collected from AMFI and related sources.

```python
files = sorted(os.listdir(folder))

for file in files:
    
    if file.endswith(".csv"):
        
        print(file)
```

## Dataset Summary

Display row and column counts for every dataset.

```python
summary = []

for file in os.listdir(folder):

    if file.endswith(".csv"):

        file_path = folder / file

        df = pd.read_csv(file_path)

        summary.append({
            "Dataset": file,
            "Rows": df.shape[0],
            "Columns": df.shape[1]
        })

summary_df = pd.DataFrame(summary)

summary_df
```

## Data Profiling

For each dataset:

- Shape
- Columns
- Missing Values
- Duplicate Records

```python
for file in os.listdir(folder):

    if file.endswith(".csv"):

        file_path = folder / file

        df = pd.read_csv(file_path)

        print("\n" + "="*60)
        print(file)

        print("Shape:", df.shape)

        print("Columns:")
        print(list(df.columns))

        print("\nMissing Values:")
        print(df.isnull().sum())

        print("\nDuplicates:")
        print(df.duplicated().sum())
```

## Dataset Preview

Inspect the first five rows of each dataset.

```python
for file in os.listdir(folder):

    if file.endswith(".csv"):

        print("\n" + "="*60)
        print(file)

        df = pd.read_csv(folder / file)

        display(df.head())
```
